- Run 1 evaluates `12 + 5`, updating `readings` to 17.

Because the notebook remembers the modified state in the active kernel, repeated execution causes cumulative increments. To ensure reproducibility and reset the variable to its expected state (17), the runtime must be restarted and cells executed linearly from top to bottom.
